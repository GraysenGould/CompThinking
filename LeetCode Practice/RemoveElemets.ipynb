```python
#given
nums = [0,1,2,2,3,0,4,2]
val = 2

#solution
i = 0
length = len(nums)

for j in range(len(nums)):
    if nums[j] != val:
        nums[i] = nums[j]
        i += 1

#for j in range(length - i):
#    nums.pop(-1)

while length > i:
    nums.pop(length - 1)
    length -= 1 


print(nums)

#soution was accepted, top 100 percent solution!!!!!
```
