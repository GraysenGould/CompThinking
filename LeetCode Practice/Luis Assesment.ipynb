```python
def fizzbuzz(n):
    for i in range(1, n +1):
        if (i % 3 == 0) and not (i % 5):
            print("FizzBuzz")
        elif (i %3 == 0):
            print("Fizz")
        elif (i % 5 == 0):
            print("Buzz")
        else:
            print(i)
        
fizzbuzz(100)
```

```python
device_names = ["switch", "tv", "switch", "tv", "switch", "tv"]
def name_items (device_names):
    new_names = []
    
    for name in device_names:
        if name not in new_names:
            new_names.append(name)
        else:
            n = 1
            test_name = name + str(n)
            while test_name in new_names:
                n += 1
                test_name = name + str(n)
            new_names.append(test_name)
    return new_names
            
name_items(device_names)
```

```python
colors = "wwwbbbbwww"
def gameWinner (colors):
    bob_moves = 0
    wendy_moves = 0
    for i in range(1, len(colors) - 1):
        if colors[i] == "w" and colors[i-1] =="w" and colors[i+1] == "w":
            wendy_moves += 1
        if colors[i] == "b" and colors[i-1] == "b" and colors[i+1] == "b":
            bob_moves += 1
    if bob_moves == 0 and wendy_moves == 0:
        return "bob"
    elif bob_moves > wendy_moves:
        return "wendy"
    else:
        return "bob"
gameWinner(colors)
```
