```python
# more elegant

def findMaxAverage (nums, k):
    current = sum(nums[:k])
    maximum = current
    for i in range(k, len(nums)):
        current += nums[i]-nums[i - k]
        if current > maximum:
            maximum = current
    return maximum / k

case = case1
findMaxAverage(case["nums"], case["k"])
```

```python
def findMaxAverage (nums, k):
    run_sum = sum(nums[:k])
    max_sum = run_sum
    
    for r in range(k, len(nums)):
        run_sum += nums[r] - nums[r-k]
        if run_sum > max_sum:
            max_sum = run_sum
    return max_sum / k

case = case1
findMaxAverage(case["nums"], case["k"])
```
