## Neetcode Solution

```python
# most efficient solution generally
def group_anagrams(strs):
    res = dict()

    for s in strs:
        count = [0] * 26
        for c in s:
            count[ord(c) - ord("a")] += 1
            
        count = tuple(count)
        if not res.get(count):
            res[count] = []
        res[count].append(s)
            
    return list(res.values())
```

```python
# most efficient solution in python
def group_anagrams(strs):
    res = dict()

    for s in strs:
        sorted_str = ''.join(sorted(s))
        if not res.get(sorted_str):
            res[sorted_str] = []
        res[sorted_str].append(s)
            
    return list(res.values())
```
