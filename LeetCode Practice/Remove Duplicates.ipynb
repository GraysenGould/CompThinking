```python
def removeDup (nums):
    k = 1
    
    for i in range(1,len(nums)):
        if nums[i] != nums[i -1]:
            if nums[k] != nums[i]:
                nums[k] = nums[i]
            k += 1

    for i in range(k,len(nums))[::-1]:
        nums.pop(i)
    return(nums)
```
