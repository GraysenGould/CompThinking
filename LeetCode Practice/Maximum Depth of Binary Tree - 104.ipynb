```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
    def __str__ (self):
        return f"val: {self.val}, left: {self.left}, right: {self.right}"
```

```python
# dumb down solution
root = [3,9,20,None,None,15,7]

binary_tree = TreeNode(3)
binary_tree.left = TreeNode(9)
binary_tree.left.left = None
binary_tree.left.right = None
binary_tree.right = TreeNode(20)
binary_tree.right.left = TreeNode(15)
binary_tree.right.right = TreeNode(7)
```

```python
def get_depth (node):
    if not node:
        return 0
    return 1 + max(get_depth(node.left), get_depth(node.right))
get_depth(binary_tree)
```
