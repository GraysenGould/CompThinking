```python
#possibilities

#entire string does not repeat, very long
#empty string
#repeat at end 
#repeat in beginning 
# repeat in the middle
```

```python
def get_no_repeat (s):
    longest = ""
    found = []
    k = 0

    for i in range(len(s)):
        if s[i] not in found:
            found.append(s[i])
        elif len(longest) < i - k +1:
            longest = s[k:i]
            k = i
    return longest
    


get_no_repeat(case4)
```

### Try 2 - My most efficient solution - O(n^2)

```python
case1 = "abcabcbb" #3
case2 = "bbbbb" #1
case3 = "pwwkew" #3
case4 = "hkelkmp" #5
case5 = "" #0
case6 = "b" #1
cases = [case1, case2, case3, case4, case5, case6]
```

```python
def get_substring (s):
    length = len(s)
    l, r = 0, 1
    longest = 0
    
    for r in range(length):
        #print(s[l:r + 1])
        for i in range(l, r):
            if s[r] == s[i]:
                l = i +1
        if r - l +1 > longest:
            longest = r - l +1
            
    return longest
    
            

get_substring("abcabcbb")
```

```python
for x in cases:
    print(get_substring(x))
```

# Neetcode Solution - O(n)

```python
def length (s):
    charSet = set()
    l = 0
    res = 0

    for r in range(len(s)):
        while s[r] in charSet:
            charSet.remove(s[l])
            l += 1
        charSet.add(s[r])
        res = max(res, r-l +1)
    return res
length(case1)
```

```python
for x in cases:
    print(length(x))
```
