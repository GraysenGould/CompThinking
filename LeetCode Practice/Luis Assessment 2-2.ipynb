```python
name = ["ball", "bat", "glove", "glove", "glove"]
price = [2, 3, 1, 2, 1]
weight = [2, 5, 1, 1, 1]

def count_unique(name, price, weight):
    unique_products = set()
    
    for i in range(len(name)):
        key = (name[i], price[i], weight[i])
        unique_products.add(key)
    
    return len(unique_products)

print(count_unique(name, price, weight))
```
