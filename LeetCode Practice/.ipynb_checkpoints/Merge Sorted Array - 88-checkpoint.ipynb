```python
nums1 = [1,2,3,0,0,0]; nums2 = [2,5,6]

nums1.extend(nums2)
new_list = [x for x in nums1 if x != 0]
nums1.clear()
for i in new_list:
    nums1.append(i)
nums1.sort()
print(nums1)
```

```python
nums1 = [1,2,3,0,0,0]; nums2 = [2,5,6]
nums3 = nums1.copy()
print(nums3)
m = 1; n = 0;
mp = 0
np = 0

for i in range(m+n):
    print(f"i = {i}, mp = {mp}, np = {np}")
    print(f"{nums3[mp]} > {nums2[np]}")
    if nums3[mp] < nums2[np] and mp < 3:
        nums1[i] = nums3[mp]
        mp += 1
    else:
        nums1[i] = nums2[np]
        np += 1
        
print(nums1)
```

```python
# try 2
nums1 = [1,2,3,0,0,0]; nums2 = [2,5,6]
nums3 = nums1.comy()
m = 3; n = 3;
i,j,k = 0, 0, 0

while i < m and j < n:
    if nums3[i] < nums2[j]:
        nums1[k] = nums3[i]
        i += 1
        k += 1
        
    
```

# NeetCode solution

```python
nums1 = [1,2,3,0,0,0]; nums2 = [2,5,6]
m = 3; n = 3;

last = m+n - 1

# merge in reverse order

while m > 0 and n > 0:
    if nums1[m - 1] > nums2[n - 1]:
        nums1[last] = nums1[m - 1]
        m -= 1
    else:
        nums1[last] = nums2[n - 1]
        n -= 1
    last -= 1

while n > 0:
    nums1[last] = nums2[n - 1]
    n, last = n - 1, last - 1
print(nums1)

```
