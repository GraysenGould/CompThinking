```python
case1 = "A man, a plan, a canal: Panama"
case2 = "race a car"
case3 = ""
case4 = "0P"
```

```python
#sloppyest method
new_str = ""
s = case4

#format string
for i in s:
    if i.isalpha() or i.isnumeric():
        new_str = new_str + i.lower() 


# check if palindrome
length = len(new_str)
palindrome = True
for i in range(length):
    if i >= (length - 1 -i):
        break
    print(f"{new_str[i]} != {new_str[length - 1 - i]}")
    if new_str[i] != new_str[length - 1 - i]:
        palindrome = False
        break
palindrome
```

```python
#most efficient solution
s = case1
s = [x.lower() for x in s if x.isalnum()]
s == s[::-1]
```

## NeetCode solution

```python
case1 = "A man, a plan, a canal: Panama"
case2 = "race a car"
case3 = ""
case4 = "0P"
```

```python
s = case1
#"cheeting solution"
newStr = "" 

for c in s:
    if c.isalnum():
        newStr += c.lower()
newStr == newStr[::-1]
```

```python
#nums: 48 to 57
#uppercase: 65 to 90
#lowercase: 97 to 122

#helper function
def alphaNum(c):
    return (ord("A") <= ord(c) <= ord("Z")
    or ord("a") <= ord(c) <= ord("z")
    or ord("0") <= ord(c) <= ord("9"))

def isPalindrome(s):
    l, r = 0, len(s) -1
    
    while l < r:
        while l < r and not alphaNum(s[l]):
            l += 1
        while r > l and not alphaNum(s[r]):
            r -= 1
        if s[l].lower() != s[r].lower():
            return False

        l, r = l+1, r -1
    return True
        
isPalindrome(" ")
```
