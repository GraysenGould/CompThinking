```python
colors = "wwwbbbbwww"


def gameWinner (colors):
    bob_moves = 0
    wendy_moves = 0
    for i in range(1, len(colors) - 1):
        if colors[i] == "w" and colors[i-1] =="w" and colors[i+1] == "w":
            wendy_moves += 1
        if colors[i] == "b" and colors[i-1] == "b" and colors[i+1] == "b":
            bob_moves += 1
    print(f"Wendy: {wendy_moves}, Bob: {bob_moves}")
    
    if bob_moves > wendy_moves:
        return "bob"
    else:
        return "wendy"
gameWinner(colors)
```
