```python
case1 = "abcabcbb"
case2 = "bbbbb"
case3 = "pwwkew"
case4 = "hkelkmp"
```

```python
#possibilities

#entire string does not repeat, very long
#empty string
#repeat at end 
#repeat in beginning 
# repeat in the middle
```

```python
def get_no_repeat (s):
    longest = ""
    found = []
    k = 0

    for i in range(len(s)):
        if s[i] not in found:
            found.append(s[i])
        elif len(longest) < i - k +1:
            longest = s[k:i]
            k = i
    return longest
    


get_no_repeat(case4)
```
