```python
#use sliding window approach - may work - "inchworm approach"
# use a while look
# min_size should start at len(arary)
# have two poiners, l and r (left and right)
# keep a running sum, starting with sum = array[l]
# if sum is smaller than target, increase l by 1, recalculate sum by adding array[l]
# check if array is smaller than min_size, if so, set min_size = l - r +1
# is sum is larger than target, increase r by 1, recalculate sum by sumbtracting array[r]
# check if array is smaller again
# stop while loop when l = len(array) - 1



```

```python
def minSubArrayLen (target, nums):
    min_size = len(nums)
    r, l = 0, 0
    running_sum = nums[0]
    while r < len(nums) - 1:
        
        print(r,l)
        if running_sum < target or l == r:
            running_sum += nums[l + 1]
            r += 1
        elif running_sum > target:
            running_sum -= nums[l + 1]
            l += 1
        if (l - r + 1) < min_size and running_sum >= target:
            print(f"new_size = {r - l +1}, {min_size}")
            min_size = l - r +1
    if min_size == len(nums):
        return 0
    return min_size
            
```
