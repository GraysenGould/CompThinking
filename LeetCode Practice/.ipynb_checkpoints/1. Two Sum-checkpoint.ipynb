```python
case1 = {"nums": [2,7,11,15], "target": 9} # [0,1]
case2 = {"nums": [3,2,4], "target": 6} # [1,2]
case3 = {"nums": [3,3], "target":  6} # [0,1]
case4 = {"nums": [1,2,2,2,3,3], "target": 6} # [4,5]
```

```python
# every input has exacly one solution

# edge cases:
# numbers needed are both first numbers
# numbers needed are both last numbers
# one number is first element, second is last
```

```python
# brute force:
# go to every element
# go through every other element in remainder of list
# check to see if numbers add up to target
# if so, return positions
# complexity O(n^2)
```

```python
def twoSum (nums, target):
    
    for i in range(len(nums) - 1):
        for j in range(i + 1,len(nums)):
            if nums[i] + nums[j] == target:
                return [i,j]
```

```python
case = case4
twoSum(case["nums"], case["target"])
```

```python
# try 2 - hashmap
# convert array into hasmap (dictionary)
# look for all combinations
# find maximum and minimum 
# go from minimum to maximum, check if exists 
# if does exist, check if target - num is in dictionary
# if so, return two values

#nightmare edgecases
```

```python
def twoSum (nums, target):
    num_hash = dict()
    min_num = max_num = nums[0]
    for i,n in enumerate(nums):
        min_num = min(min_num, n)
        max_num = max(max_num, n)
        if not num_hash.get(n):
            num_hash[n] = [i]
        else:
            num_hash[n].append(i)
            
    for i in range(min_num, max_num +1):
        if num_hash.get(i) and num_hash.get(target - i):
            if len(num_hash[i]) == 2:
                return num_hash[i]
            else:
                return [num_hash[i][0], num_hash[target - i][0]]
    
        
```

```python
case = case2
twoSum(case["nums"], case["target"])
```

```python
#third try
def twoSum (nums, target):
    num_hash = dict()
    min_num = max_num = nums[0]
    for i,n in enumerate(nums):
        min_num = min(min_num, n)
        max_num = max(max_num, n)
        if not num_hash.get(n):
            num_hash[n] = [i]
        else:
            num_hash[n].append(i)
    for n in num_hash:
        if num_hash.get(target - n):
            if len(num_hash[n]) == 2:
                return num_hash[n]
            elif n + n != target:
                print("yippee")
                return [num_hash[n][0], num_hash[target - n][0]] 

```

```python
case = case4
twoSum(case["nums"], case["target"])
```

```python
# best solution
def twoSum (nums, target):
    cache = {}
    for i in range(len(nums)):
        if cache.get(target - nums[i]) != None:
            return [cache[target - nums[i]], i]

        cache[nums[i]] = i
```

```python
case = case4
twoSum(case["nums"], case["target"])
```
