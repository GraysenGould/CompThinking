```python
str1 = "hackthegame" # 0
# if no unique charactors, reutrn -1
# add one to index of the first unique charactor
```

```python

def find_index (str1):
    first_occur = {}
    unique_dict = {}
    min_index = len(str1) - 1
    for i in range(len(str1)):
        if str1[i] not in first_occur:
            first_occur[str1[i]] = i
            unique_dict[str1[i]] = str1[i]
        elif str1[i] in first_occur and str1[i] in unique_dict[str1[i]]:
            del unique_dict[str1[i]]
    for c in unique_dict:
        min_index = min(first_occur[c], min_index)
    
    return min_index + 1

str1 = "hackthegame"
print(find_index(str1))
```

```python
name = ["ball", "bat", "glove", "glove", "glove"]
price = [2,3,1,2,1]
weight = [2,5,1,1,1]

def count_unique (name, price, weight):
    unique_count = {}
    count = 0
    
    for i in range(len(name)):
        key = (name[i], price[i], weight[i])
        
        if key not in unique_count:
            unique_count[key] = 1
        else:
            unique_count[key] += 1
    
    for p in unique_count:
        if unique_count[p] != 1:
            count += 1
    return count
    
print(count_unique(name, price, weight))
```
