```python
case1 = [7,1,5,3,6,4]
case2 = [7,6,4,3,1]
case3 = [2,1,2,1,0,1,2]
```

```python
#possibilities:
# array is size of 1
# no buy and sell dates make profit (case2)
# array is size of 2
# prices is higher day 1 than day 2, cannot buy day 2, sell day 1
```

### brute force

```python
def max_profit(prices):
    max_profit_by_day = []
    for i in range(len(prices)): 
        for p2 in prices[i:]:
            max_profit_by_day.append(p2 - prices[i])
    print(max_profit_by_day)
    return max(max_profit_by_day)

max_profit(case1)
```

# better solution

```python
def max_profit (prices):
    length = len(prices)
    i, f = 0,1
    best_deal = 0
    new_i = 0
    while i < length:
        #print(f"first round {i}")
        
        for f, p in enumerate(prices[i:]):
            #print(f"second round {f}")
            #print(f" i = {i}, f = {f}")
            if p - prices[i] > best_deal:
                best_deal = p - prices[i]
            if p < prices[i]:
                print(f"initial price: {prices[i]}, final price: {p}, days: {(i, f)}")
                new_i = f + i - 1
                pass
        if new_i != i:
            i = new_i
        i += 1
    return best_deal
max_profit(case3)
# [2,1,2,1,0,1,2]
```

### using sliding window

```python
case1 = [7,1,5,3,6,4]
case2 = [7,6,4,3,1]
case3 = [2,1,2,1,0,1,2]
```

```python
def max_profit (prices):
    l = 0
    most_made = 0
    for r in range(len(prices)):
        if prices[r] - prices[l] > most_made:
            most_made  = prices[r] - prices[l]
        if prices[r] < prices[l]:
            l = r
    return most_made

max_profit(case3)
```
