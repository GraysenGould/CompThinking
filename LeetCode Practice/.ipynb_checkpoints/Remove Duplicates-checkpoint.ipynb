#### possiblilites
1. no duplicates
2. only duplicates
3. empty array - should not be possible
4. one element array
5. duplicates only at beginnning
6. duplicates only at end

```python
tests = []

tests.append({
    "input":{"nums": [1,2,3,4]},
    "output": [1,2,3,4]
})
tests.append({
    "input":{"nums": [1,1,1,1]},
    "output": [1]
})
tests.append({
    "input":{"nums": [1]},
    "output": [1]
})
tests.append({
    "input":{"nums": [1,1,1,4,5,6]},
    "output": [1,4,5,6]
})
tests.append({
    "input":{"nums": [1,2,3,4,5,5,5,5,5]},
    "output": [1,2,3,4,5]
})
print(tests)
```

```python
def removeDuplicates (nums):
    k = 0
    prev_val = nums[0]

    for i in range(1, len(nums)):
        print(f"debug: i = {i}, k = {k}, nums = {nums}, prev_val = {prev_val}")
        if nums[i] != prev_val:
            print("moving as normal")
            k += 1
            prev_val = nums[i]

        elif nums[i] != nums[k] and nums[i-1] == nums[k]:
            print(f"happened")
            nums[k+1] = nums[i]
            k+=1
        else:
            print("what!!!")
        
        
    return nums
```

```python
print(removeDuplicates([1,2,2,4]))
```
