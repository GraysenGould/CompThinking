```python
#make it more efficient

def maximumSubarraySum (nums, k):
    def add_val (count_dict, val):
        if val not in count_dict:
            count_dict[val] = 1
        else:
            count_dict[val] += 1
    
    max_sum  = 0
    sub_array = nums[:k-1]
    run_sum = sum(sub_array)
    length = len(nums)
    l,r = 0, k-2
    count_dict = {}
    
    for n in nums[:k-1]:
        add_val(count_dict, n)
        
    for r in range(k - 1, length):

        run_sum += nums[r]
        sub_array.append(nums[r])
        add_val(count_dict, nums[r])
    
        unique = True
        for n in count_dict:
            if count_dict[n] > 1:
                unique = False
        
        if unique and run_sum > max_sum:
            max_sum = run_sum

        count_dict[nums[l]] -= 1
        
        run_sum -= nums[l]
        sub_array.pop(0)
        l += 1
        
    return max_sum

case1 = {"nums":[1,5,4,2,9,9,9], "k":3}
case = case3
maximumSubarraySum(case["nums"], case["k"])
```

```python
#better
from collections import defaultdict
def maximumSubarraySum (nums, k):
    max_sum = 0
    window_sum = 0
    start = 0
    n = len(nums)
    count_map = defaultdict(int)

    for i in range(n):
        window_sum += nums[i]
        count_map[nums[i]] += 1

        if i - start + 1 == k:
            if len(count_map) == k:
                max_sum = max(max_sum, window_sum)
            window_sum -= nums[start]
            count_map[nums[start]] -= 1
            
            if count_map[nums[start]] == 0:
                del count_map[nums[start]]
            start += 1
    return max_sum


     
case1 = {"nums":[1,5,4,2,9,9,9], "k":3}
case = case3
maximumSubarraySum(case["nums"], case["k"])
```
