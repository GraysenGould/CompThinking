Write a script to explore the above array for a treasure. The values in the array are clues. Each cell contains an integer between 11 and 55; for each value the ten's digit represents the row number and the unit's digit represents the column number of the cell containing the next clue. Starting in the upper left corner (at 1,1), use the clues to guide your search of the array. (The first three steps are 11 -> 34 -> 42). The treasure is a cell whose value is the same as its coordinates. Your program must first read in the treasure map data into a 5 by 5 array. The treasure map is located at [http://54.243.252.9/engr-1330-webroot/8-Labs/Lab07/treasure1.txt](http://54.243.252.9/engr-1330-webroot/8-Labs/Lab07/treasure1.txt).
Your program should output the cells it visits during its search, and a message indicating where you found the treasure.

```python
# put your algorithm here

# objective: find elements in map where value is equal to corrdinates 
# 1: create an array for 
# 2: start in first cell check if coordinates equal value
# 3: else, add position to list, go to the new position, repeat step 2
# 4: If value of corrdinates is the same as the coordinates, break the loop and output the list of corrdinates
# 5: for simplicity, if length of array of positions is greater than the map size, exit loop and report no treasure
```

```python
import pandas as pd

# create storage for the map
data = [[34, 21, 32, 41, 25],
        [14, 42, 43, 14, 31],
        [54, 45, 52, 42, 23],
        [33, 15, 51, 31, 35],
        [21, 52, 33, 13, 23]]
pd.DataFrame(data).to_csv("treasure_map.csv", index = False, header = False)

# read in the data from the file
treasure_map = pd.read_csv("treasure_map.csv", header = None)


def find_treasure (treasure_map):
    pos = [1, 1] # start at 1,1 and follow the clues
    map_size = treasure_map.size
    visited = []
    
    while len(visited) <= map_size:
        value = str(treasure_map.iloc[pos[0] - 1, pos[1] - 1]) # find value of current cell

        visited.append(pos)     
        if int(value[0]) == pos[0] and int(value[1]) == pos[1]: # detect if treasure found
            
            return "treasure found", visited # report treasure found
        else:
            pos = [int(value[0]), int(value[1])]
            if pos in visited: # if a position is repeated, we are in a loop, no treasure
                return "no treasure", visited # report if no treasure found
    return visited
            
result = find_treasure (treasure_map)
print(f"{result[0]} \n{result[1]}")
```
