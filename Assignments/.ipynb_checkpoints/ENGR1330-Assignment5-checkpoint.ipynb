## Question 3 (20 pts):
Find the root of the function $y\left(x\right) = x^3-2x-5x$ using the bisection method and the secant method. Plot the function and the approximations from both methods.

```python
# did you mean for the function to be x^3 -2x^2 -5x ??
```

```python
# Question 3 Code Bisection Goes Here
def find_root_bisection (y, a, b):
    interval = [a, b] # initial guesses
    error = 1 # start with an arbitrary error greater than max_error
    max_error = 1e-5
    iterations = 0
    
    # while loop to calculate the root of function y
    while error > max_error:
        iterations += 1
        guess = sum(interval)/2
        if y(interval[0])*y(guess) > 0: # meaning sign is same as left interval
            interval[0] = guess
        else:
            interval[1] = guess
        error = abs(interval[1] - interval[0]) / guess * 100
        
    print(f"Root found in {iterations} iterations.")
    return guess

y = lambda x: x**3 - 2*x - 5*x # function: x^3 - 2x - 5x = x^3 - 7x

root = find_root_bisection(y, 1, 5) # start with guesses 1 and 5

print(f"The root of this function is: {round(root, 5)}")
```

```python
# Question 3 Code Secant Goes Here
#x2 = x1 - f(x1)(x1-x0)/ (f(x1) - f(x0))

def find_root_secant (y, guess1, guess2):
    error = 1 # start with an arbitrary error greater than max_error
    max_error = 1e-5
    iterations = 0
    x0, x1 = guess1, guess2

    while error > max_error:
        iterations += 1
        new_guess = x1 - y(x1)*(x1- x0) / (y(x1) - y(x0)) # create new guess based on formula

        # shift around some variables
        x0 = x1
        x1 = new_guess


        error = abs((x1 - x0) / x1) * 100 # calculate error
        
    print(f"Root found in {iterations} iterations.")
    return new_guess

y = lambda x: x**3 - 2*x - 5*x # function: x^3 - 2x - 5x = x^3 - 7x

root = find_root_secant(y, 1, 5) # start with guesses 1 and 5
print(f"The root of this function is: {round(root, 5)}")
```

## Question 4 (20 pts):
You will simulate heat conduction along a metal rod using the finite difference method. The goal is to observe how the temperature evolves over time.

### Given:
- **Length of the rod \(L\)**: 1 meter
- **Initial temperature distribution**: Uniform at 100°C
- **Boundary conditions**: 
  - \(u(0, t) = 100\) °C (left end)
  - \(u(L, t) = 0\) °C (right end)
- **Total simulation time \(T\)**: 1 second
- **Spatial step \( Delta x\)**: 0.1 m
- **Time step \( Delta t\)**: 0.01 s
- **Thermal diffusivity \(alpha)**: 0.1 m²/s

## Tasks

1. Initialize the temperature distribution along the rod.
2. Use the explicit finite difference method to update the temperature at each time step.
3. Apply the boundary conditions at each time step.
4. Visualize the temperature distribution at selected time intervals.

   Use the below example code as an outline for your code

```python
# EXAMPLE CODE STRUCTURE- USe it to write yor own below
import numpy as np
import matplotlib.pyplot as plt

# Parameters
L = 1.0           # Length of the rod
alpha = 0.1       # Thermal diffusivity
T = 1.0           # Total time
dx = 0.1          # Spatial step
dt = 0.01         # Time step

# Discretization
x = np.arange(0, L + dx, dx)
Nx = len(x)
time_steps = int(T / dt)

# Initial condition
u = np.ones(Nx) * 100  # Initial temperature: 100°C

# Finite difference method
for n in range(time_steps):
    u_new = np.copy(u)
    for i in range(1, Nx - 1):
        u_new[i] = u[i] + alpha * dt / dx**2 * (u[i-1] - 2*u[i] + u[i+1])
    
    # Apply boundary conditions
    u_new[0] = 100  # Left end
    u_new[-1] = 0   # Right end
    
    u = u_new

    # Visualization (plot every 10 steps)
    if n % 10 == 0:
        plt.plot(x, u, label=f'Time = {n*dt:.2f}s')

# Final plot settings
plt.title('Temperature Distribution along the Rod')
plt.xlabel('Position along the rod (m)')
plt.ylabel('Temperature (°C)')
plt.legend()
plt.grid()
plt.show()
```

```python
# Question 4 Code Goes Here
import numpy as np
import matplotlib.pyplot as plt

# setting up variables
l = 1
dx = 0.1
temp_i = 100
time = 1
dt = 0.01
alpha = 0.1

x = np.arange(0, l + dx, dx)
time_axis = np.arange(0, time, dt)

u = np.ones(len(x)) * temp_i

plt.figure(figsize = (8, 6)) # set up plot figure


for n in range(len(time_axis)):
    u_new = np.copy(u)

    # calculate new temperatures
    for i in range(1, len(x) - 1):
        u_new[i] = u[i] + alpha *dt / dx**2 * (u[i - 1] - 2*u[i] + u[i+1])
        
    u_new[0] = 100 
    u_new[-1] = 0

    u = u_new

    # plot temperatures every 10 iterations
    if n % 10 == 0:
        plt.plot(x, u, label = f"t = {round(dt * n, 1)}s")


# format and add labels to chart        
plt.title("Temperature Across Metal Bar", fontsize = 20)
plt.xlabel("Position Along Bar (m)")
plt.ylabel("Temeperature (°C)")
plt.legend()
plt.show()
    
```
