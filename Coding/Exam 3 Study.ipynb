```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
players = pd.read_csv("player_data.csv")
```

```python
players.head()
```

```python
salaries = players["Salary"]
salary_mean = round(salaries.mean())
salary_mean
```

```python
sample_means = []
n = 1000
for i in range(n):
    
    sample = salaries.sample(1000, replace = True)
    mean = sample.mean()
    sample_means.append(mean)

overall_mean = round(np.mean(sample_means))

plt.hist(sample_means)
plt.scatter(overall_mean, -5, color = "red")
plt.show()
print(f"The overall mean is {overall_mean}")
```

```python
error = round((overall_mean - salary_mean) / salary_mean * 100, 2)
error
```
