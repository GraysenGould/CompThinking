```python
#Temperature Problem
temp_data = np.random.randint(2, 10, (2, 10))
temp_data_f = temp_data*1.8 + 32

avg_temp = np.mean(temp_data_f, axis = 1)
sensor_idx = np.argmax(avg_temp)


k = 3
pad = (k//2, k//2)
temp_data_f_padded = np.pad(temp_data_f, [(0,0), pad], mode = 'edge')
print(temp_data_f)
print(temp_data_f_padded)

kernel = np.ones(k) / k
temp_data_f_moving_avg = np.apply_along_axis(lambda x: np.convolve(x, kernel, mode='valid'), axis=1, arr=temp_data_f_padded)

print(temp_data_f_moving_avg)
```

## Linear Algebra Numpy

```python
import numpy as np
from numpy.linalg import norm
```

```python
# column and row vectors:
c_vec = np.array([[1,6,7,3]])
r_vec = np.array([[1], [2],[5], [6]])

print(c_vec,'\n\n', r_vec)
```

```python
new_vec = c_vec.T
print(new_vec)
```

```python
norm_1_1 = norm(c_vec, 1)
norm_1_2 = norm(c_vec, 2)
norm_1_inf = norm(c_vec, np.inf,)

print(f"order of 1: {norm_1_1}, order of 2: {norm_1_2}, order of infinity: {norm_1_inf}")
```

```python
# finding theta:
arr1 = np.array([[1,4,3]])
arr2 = np.array([[4,-9,7]])
arr2_t = arr2.T

print(np.cross(arr1, arr2))
```
