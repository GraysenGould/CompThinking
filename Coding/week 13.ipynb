```python
nums1 = [1,5,4,2,9,9,9]
k1= 3
```

```python
def maximumSubarraySum (nums, k):
    run_sum = 0
    max_sum = 0 # need to change 
    length = len(nums)
    unique = set()
    l = 0
    for r in range(0, length):
        while nums[r] in unique or (r - l + 1 > k):
            unique.remove(nums[l])
            run_sum -= nums[l]
            l += 1        
        run_sum += nums[r]
        unique.add(nums[r])
        
        if r - l + 1 == k and run_sum > max_sum:
            max_sum = run_sum
    return max_sum
    
maximumSubarraySum(nums1, k1)
#lets go!!! it works!!!
```
