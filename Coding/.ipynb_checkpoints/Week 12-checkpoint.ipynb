### sampling and probalility

```python
# use df.sample(100, replace = False) to get a sample of a dataset
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
rolls = np.arange(1,7)

results = []
results.append(np.random.choice(rolls, 1000))

plt.hist(results, bins = 6)
```

```python
# white, red, green ball problem
def color_sim (sim_num):
    colors = np.array(["white", "red", "green"])

    case1 = 0
    case2 = 0
    for i in range(sim_num):

        picked = np.random.choice(colors, 5, p = [1/6, 1/3, 1/2], replace = True)
        if sum(picked == "white") == 3 and sum(picked == "red") == 2:
            case1 += 1
        if sum(picked == "white") == 5 or sum(picked == "red") == 5 or sum(picked == "green") == 5:
            case2 += 1
    print(case1)
    prob1 = round((case1/sim_num) * 100, 3)
    prob2 = round((case2/sim_num) * 100, 3)
    return prob1, prob2
    
color_sim(100000)
```
