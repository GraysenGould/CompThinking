```python
import numpy as np
import pandas as pd
```

```python
products = pd.DataFrame({"product_id": [0,1], "store1": [95,70], "store2": [100, None], "store3": [105, 80]})
```

```python
products
```

```python
#products2 = pd.DataFrame([[0,95], [1,70]], columns = ["product_id", "store1"])
#note the other way to use a dataframe
```

```python
new_products = products[["product_id", "store1"]]
new_products.insert(1, "store", ["store1", "store1"])
new_products.rename(columns = {"store1": "price"})
```

```python
new_products2 = products[["product_id", "store2"]]
new_products2.insert(1, "store", ["store2", "store2"])
new_products2.rename(columns = {"store2": "price"})
```

```python
print(pd.concat([new_products, new_products2]))
```

```python
store_list = []
for i in products.columns[1:]:
    new_products = products[["product_id", i]].copy()
    new_products.insert(1, "store", np.repeat(i, len(new_products))) # make more versitile later
    new_products = new_products.rename(columns = {i: "price"})
    store_list.append(new_products)

new_table = pd.concat(store_list)
new_table.dropna(inplace = True)
new_table
```
