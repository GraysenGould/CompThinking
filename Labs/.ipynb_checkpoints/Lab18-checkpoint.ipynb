### Example: From a normally distributed population, we randolmy took a sample of 500 students with a mean score of 461 on the math section of SAT. Suppose the standard deviation of the population is 100, what is the estimated true population mean for the 95% confidence interval.

```python
ci = 0.95 #how can we get zscore
```

![](https://www.w3schools.com/statistics/img_standard_normal.svg)

```python
#Our confidence interval is always the "middle" that is two-tail
(1 - 0.95)/2
```

```python
from scipy.stats import norm
norm.ppf(0.025)
```

```python
# Step 1- Organize the data
ss =500
xbar = 461
std = 100
z_score = 1.96
```

```python
# Assuming a normally distributed population
# Assuming randomly selected samples
# Step2- Calculate the margin of error
standard_error = std/(ss**0.5)
margin_error = z_score * standard_error
```

```python
# Step3- Find the estimated true population mean for the 95% confidence interval
# To find the range of values you just have to add and subtract ??? from ???
lower = round(xbar - margin_error)
upper = round(xbar + margin_error)
print(f"The 95% confidence interval is {lower}, {upper}")
```

```python
round(margin_error)
```

### What is the estimated true population mean for the 90% confidence interval?

```python
ss =500
xbar = 461
std = 100
ci = .90
z_score = round(abs(norm.ppf((1-ci)/2)),4)
standard_error = std / (ss**0.5)
moe = z_score * standard_error
lower = round(xbar - moe)
upper = round(xbar + moe)
print(f"The 90% confidence interval is {lower}, {upper}")
```

### What is the estimated true population mean for the 99% confidence interval?

```python
ss =500
xbar = 461
std = 100
ci = .99
z_score = round(abs(norm.ppf((1-ci)/2)),4)
standard_error = std / (ss**0.5)
moe = z_score * standard_error
lower = round(xbar - moe)
upper = round(xbar + moe)
print(f"The 99% confidence interval is {lower}, {upper}")
```

### What is the relationship between the confidence interval and margin of error?

```python
# the more confident we are (say 99%), the larger the larger our margin of error
# the less confident we are (say 90%), the smaller the margin of error we have 
# in other words, there is a tradeoff between how confident we can be and how small our margin of error is.
```

### Example: Solve the above problem assuming that we have samples from 1000 students.

```python
ss = 1000
xbar = 461
std = 100
```

### What is the estimated true population mean for the 90% confidence interval?

```python
ci = .90
z_score = round(abs(norm.ppf((1-ci)/2)),4)
standard_error = std / (ss**0.5)
moe = z_score * standard_error
lower = round(xbar - moe)
upper = round(xbar + moe)
print(f"The 90% confidence interval is with 1000 population size is {lower}, {upper}")
```

### What is the estimated true population mean for the 99% confidence interval?

```python
ci = .99
z_score = round(abs(norm.ppf((1-ci)/2)),4)
standard_error = std / (ss**0.5)
moe = z_score * standard_error
lower = round(xbar - moe)
upper = round(xbar + moe)
print(f"The 90% confidence interval is with 1000 population size is {lower}, {upper}")
```

### What is the relationship between the sample size and margin of error?

```python
# the larger our sample size, the smaller our margin of error is. 
# this is because the margin of error is inversly proportional to the square root of the population size
# moe = zscore * std / sqrt(sample size)
```

### What is the relationship between the sample size and confidence interval?

```python
# the larger our sample size, the narrower our confidence interval is 
# we can be more certain (say 99%), and still get a smaller confidence interval
# this means that if we have a larger sample size, we can predict with more accuracy the mean of the data. 
```

### Example: Solve the above problem assuming that the standard deviation of the population is 50.

```python
ss = 500 # resetting population size to 500
xbar = 461
std = 50 # std is now 50
```

```python
# starting with 99% confidence
ci = .99
z_score = round(abs(norm.ppf((1-ci)/2)),4)
standard_error = std / (ss**0.5)
moe = z_score * standard_error
lower = round(xbar - moe)
upper = round(xbar + moe)
print(f"The 99% confidence interval is with a Standard Deviation of 50 is {lower}, {upper}")
```

```python
# this shows that the smaller our standard deviation, the smaller our margin of error
# in other words, if we have a small standard deviation, we can get a smaller confidence interval and better predict the mean value
```

### What is the estimated true population mean for the 90% confidence interval?

```python
# 90 percent confidence
ci = .90
z_score = round(abs(norm.ppf((1-ci)/2)),4)
standard_error = std / (ss**0.5)
moe = z_score * standard_error
lower = round(xbar - moe)
upper = round(xbar + moe)
print(f"The 90 confidence interval is with a Standard Deviation of 50 is {lower}, {upper}")
```
