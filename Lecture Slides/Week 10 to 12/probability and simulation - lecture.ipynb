# Probability and Simulation

```python
import numpy as np
import matplotlib.pyplot as plt
```

### Probability

**Definition:**

Probability is the branch of mathematics concerning numerical descriptions of how likely an is to occur, or how likely it is that a proposition is true. The probability of an event is a number between 0 and 1, where, roughly speaking, 0 indicates impossibility of the event and 1 indicates certainty. The higher the probability of an event, the more likely it is that the event will occur. 

A simple example is the tossing of a fair (unbiased) coin. Since the coin is fair, the two outcomes ("heads" and "tails") are both equally probable; the probability of "heads" equals the probability of "tails"; and since no other outcomes are possible, the probability of either "heads" or "tails" is 1/2 (which could also be written as 0.5 or 50%).

**Some examples of probability related questions we try to answer using data science:**

- Is it more likely to have higher humidity in a day that is rainy compared to a clear day?
- How likely is it to make a revenue in the range 10k-15k from a business?
- How likely it is for a patient with BMI of 32 to have diabetes?

### Simulation

<b>Example 1: Lets try to get the probability of getting a 6 from rolling of a dice using simulation.

```python
dice=[1,2,3,4,5,6]
face=[]
count=0
n=10000
for i in range(n): # simulating the dice rolling for 100 times (why we need to simulate so many times??)
    c=np.random.choice(dice)
    face.append(c)
    if c==6:
        count+=1
p=count/n
plt.hist(face,bins=6)
print("the probablity of getting a 6 from rolling of a dice is:",p)
```

### How the number of silmulations matters? 

```python
dice=[1,2,3,4,5,6]
p=[]
n=100 #number of simulations
for i in range(1,7):
    count=0
    for j in range(n): 
        c=np.random.choice(dice)
        if c==i:
            count+=1
    p.append(count/n)

plt.bar(dice,p) 
plt.show()
```

### ***The greater the number of simulations, the closer is the empirical probablity to the accurate mathematical probablity***

##### **Example 2: Imagine a box full of different colored balls, where there are 40% red, 25% blue, 20% white and 15% green. Youre supposed to randomly pick 3 balls from the box. Estimate the probability of getting all 3 of them red**

```python
box=["red","blue","green","white"] # categories present in the collection
prop=[0.4,0.25,0.15,0.2]  # proportions of different categories
target=["red","red","red"]

# simulating one experiment of picking 3 balls
ball=np.random.choice(box, size=3, p=prop)
print(ball)
```

```python
count=0

for i in range(5000):
    ball=np.random.choice(box, size=3, p=prop)
    if list(ball)==target:
        count+=1

probability=count/5000
print(probability)
```

##### **Example 3: Imagine the same box as example 2. Now youre supposed to randomly pick 3 consecutive balls from the box, with replacement. Estimate the probability of getting all 3 of them red**

```python
count=0

for i in range(500000):
    ball=np.random.choice(box, size=3, p=prop,replace=True) # note the replace argument
    if list(ball)==target:
        count+=1

probability=count/500000
print(probability)
```

## Distribution of the probability

The probability you got is one value. Lets repeat the above for 500 times. We will get 500 slightly different values of probabilities. A histogram of that will be the empirical histogram, the center of the histogram showing the accurate estimate of the probability.

```python
emp=[]

for j in range(100):
    p=0
    for i in range(1000):
        s=np.random.choice(dice)
        if s==6:
            p+=1
    emp.append(p/10000)
    
np.median(emp)
np.mean(emp)
```
