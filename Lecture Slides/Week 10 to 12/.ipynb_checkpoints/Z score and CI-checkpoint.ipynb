## Z score and confidence intervals for normal distributions

## Confidence interval

Informally, in frequentist statistics, a confidence interval (CI) is an interval which is expected to typically contain the parameter being estimated. More specifically, given a confidence level (95% and 99% are typical values), a CI is a random interval which contains the parameter being estimated % of the time. (https://en.wikipedia.org/wiki/Confidence_interval)

Also read this article: https://www.investopedia.com/terms/c/confidenceinterval.asp

**STATISTICS**

```python
#Importing libraries

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm
```

# --------------------------------------------------------------------------------------------------------------

# Estimating "Proportion"

**What proportion of parents use car seat for toddlers?**

**Out of sample of 659 parents, 540 reported they use a car seat**

**Construct a 90% confidence interval for overall population proportion**

```python
#Calculating the proportion for the given sample which will be our best estimate

prop = 540/659
prop
```

```python
#Calculating the margin of error based on the equation given for proportion (refer to slide 15 in lesson 18 for the equation)

#The value of 1.96 is the z-score for 95% CI

moe = 1.645*np.sqrt((prop * (1 - prop))/659) ## z score for 90% CI is 1.65
moe
```

```python
#Calculating the lower confidence bound and the upper confidence bound for a 95% CI as (best estimate ± margin of error)

lcb = prop - moe
ucb = prop + moe

print(lcb, ucb)
```

```python
#Calculating the lower confidence bound and the upper confidence bound for a 95% CI using the proportion_confint() function in 
#statsmodels library

prop_ci_95 = sm.stats.proportion_confint(540, 659, alpha = 0.05)
print(prop_ci_95)
```

# --------------------------------------------------------------------------------------------------------------

# Estimating "Mean"

##### **What is the population mean triple jump distance?**

**The mean triple jump distance covered by 650 athletes is 500 cm and the standard deviation is 80 cm**

**Construct a 95% interval for the population mean triple jump distance**

```python
#Storing the given mean value in a variable named 'mean'

mean = 500
mean
```

```python
#Calculating the margin of error based on the equation given for mean (refer to slide 22 in lesson 18 for the equation)

#The value of 1.645 is the z-score for 90% CI

moe = 1.96*(80/np.sqrt(650))
moe
```

```python
#Calculating the lower confidence bound and the upper confidence bound for a 90% CI as (best estimate ± margin of error)

lcb = mean - moe
ucb = mean + moe

print(lcb, ucb)
```
