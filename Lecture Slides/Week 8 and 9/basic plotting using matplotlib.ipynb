# Data visualization

Visual display of data is a powerful tool for exploratory data analysis and to present outputs of data analysis. Different type of plots:

Line plots

Bar plots

Scatter plots

Histograms


## Different plots types

***Line plots:***

Used to show how value of one variable changes in response to the changes of value of another variable
Linear or polynomial relationship between two variables. 
Also to measure trends or chronological patterns


***Bar plots:***

used to compare data values among few groups


***Scatter plots:***

used to represent how closely multiple data points lie on the X-Y plane
X and y both are independent variables 


***Histograms:***

Are created based on data values of a single variable
visual description of the data values of one variable over its own value range. Represents the statistical distribution of the data 



## Plot components:

Axes

Axes lables

Data (lines/markers)

Plot title

Legends

### Create a bar plot showing average temperatures per month for year 2022


#### Which month was hottest? And which was coolest?

#### Compare 2022 monthwise temperature data with 2021

```python
lbkw=p.read_csv("Lubbock_weather.csv")
lbkw
```

```python
temp2022=lbkw[lbkw["year"]==2022]
temp2022
```

```python
temp2021=lbkw[lbkw["year"]==2021]
temp2021
```

```python
grp22=temp2022.groupby("month")[["temp"]].mean()
grp22
```

```python
grp21=temp2021.groupby("month")[["temp"]].mean()
grp21
```

```python
finaldata1=grp21.reset_index()
finaldata2=grp22.reset_index()
finaldata2
```

```python
plt.bar(finaldata2["month"],finaldata2["temp"],color="green",width=0.5,align="edge")
plt.ylabel("Average temperature")
plt.xlabel("Month")
plt.title("Average tempertaure vs month for 2022")

plt.bar(finaldata1["month"],finaldata1["temp"],color="red",width=0.5)

plt.legend(["2022","2021"])

plt.show()
```
