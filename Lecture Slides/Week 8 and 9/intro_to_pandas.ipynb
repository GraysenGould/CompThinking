```python
# axis=1 means applying the function row-wise 
# axis=0 means defualt, column wise application of a function
# example:
test=p.DataFrame([["John",79,85,81],["Kelly",91,82,87]],columns=["Name","Exam1","Exam2","Exam3"])
test
```

```python
## apply the mean function over all columns except the first
## the iloc is used here for concvenience, to select from column 1 to 3
test.iloc[:,1:].mean(axis=1)
```

```python
# applying a function on a specific column
mydf["Age"].mean()
```

```python
mydf
```

```python
# applying a function on selected rows of a dataframe
mydf[mydf["Total Score"]>=80].mean()
```

```python
# applying a function on a specific column of the selected rows of a dataframe
mydf[mydf["Total Score"]>=80]["Total Score"].mean()
```

```python
# try this: you will get an error as the mean function only acted upon 
# Total score column!
mydf[mydf["Total Score"]>=80]["Total Score"].mean()["Age"]
```

# 5. dealing with NaN

```python
## insert some null values
test=p.DataFrame([["John",79,85,81],["Kelly",91,n.NaN,87],[n.nan,67,85,90]],columns=["Name","Exam1","Exam2","Exam3"])
```

```python
test
```

### dropping the rows/columns

```python
test
```

```python
## removing NaN data
## removing  entire row that contains at least 1 NaN
test_dropped=test.dropna() 
test_dropped
```

```python
## removing rows having NaN in specific columns

test.dropna(subset=["Exam2","Exam1"],inplace=True) 
test
```

```python
test
```

```python
test_dropped=test.dropna(axis=1) 
test_dropped
```

### replace the nulls

```python
## replace/fillup NaN with 0
test_filled1=test.fillna(0) # replace all null values wth 0
test_filled1
```

```python
test.info()
```

```python
# fill the nulls in Name column with N/A and the nulls in Exam3 with the average of the Exam3 
test_filled3=test.fillna({"Name":"N/A","Exam2":test["Exam2"].mean()}) 
# use a dictionary to provide the filler values. 
# keys of the dictionary are the columns names, values are the substitutes for nulls in respective columns
test_filled3
```

```python
test
```

```python
# replace the NaN by rowwise average. 
''''NOTE: it works fine for this case, 
 if you have more than one rows having nulls on the same column, your code will be 
 more complicated '''

test_filled3=test.fillna({"Name":"N/A","Exam2":test.iloc[1,1:].mean()})
test_filled3
```
