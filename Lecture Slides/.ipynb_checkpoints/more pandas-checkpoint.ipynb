# 1. grouping


Let's find the statewise average scores by students from the dataframe below:

```python
mydata={"Name":["Harry","Johnson","Peter","Ali","Jane","Smith","Amanda","Sophia"],"State":["TX","NY","TX","NY","NY","CA","CA","TX"],"Total Score":[82,69,98,76,88,81,90,66],"Age":[20,21,24,25,20,27,26,23]}
```

```python
mydf=p.DataFrame(mydata)
```

```python
mydf
```

```python
a=mydf.groupby("State")[["Total Score","Age"]].mean() 
a# "State" is the key column. Note: the key column become the row header in the grouped dataframe
```

```python
b=a.reset_index()
b
```

# 2. date and time

```python
timerange = p.DataFrame() ## an empty dataframe
```

```python
timerange
```

```python
# create a column and fill it with date and time data generated by date_range() function
timerange['Date and time'] = p.date_range(start='10/6/2021 10:00:32', end='12/7/2021 14:21:00', freq ='H') ## freq 'min','S','D','M','Y'
timerange
```

```python
timerange2=p.DataFrame()
timerange2['date and time']=p.date_range('10/6/2021 19:21:34',periods=100,freq='H')
timerange2
```

```python
timerange2.loc[4:10]
```

```python
timerange2['date and time'].dt.minute# day, month, hour, minute
```

# 3. Exporting dataframes

## saving dataframe as csv

```python
mydata={"Name":["Harry","Johnson","Peter","Ali","Jane","Smith","Amanda","Sophia"],"State":["TX","NY","TX","NY",n.NaN,"CA","CA",n.NaN],"Total Score":[82,69,98,76,88,81,90,66],"Age":[20,21,24,25,20,27,26,23]}
df=p.DataFrame(mydata)
```

```python
df
```

```python
df.to_csv('mydata_file.csv')  ## go to home directory nd open the csv file to check
```

```python
df1=p.read_csv("mydata_file.csv")
df1
```

```python
df1=p.read_csv("mydata_file.csv",usecols=["Name","State","Total Score","Age"],index_col="Name")
df1
```
