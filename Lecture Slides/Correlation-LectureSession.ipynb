## Lecture session for Correlation

```python
#Importing libraries

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#Reading the .csv file

hybrid = pd.read_csv('hybrid.csv')
hybrid
```

```python
#Scatter plot of price vs. acceleration of hybrid cars

plt.scatter('acceleration', 'msrp', data = hybrid, color = 'blue')
plt.xlabel('Acceleration (Km per hour per second)')
plt.ylabel('Price (2013 dollars)')
plt.title('Price vs. acceleration')
```

```python
#Scatter plot of price vs. mileage of hybrid cars

plt.scatter('mpg', 'msrp', data = hybrid, color = 'red')
plt.xlabel('Mileage (miles per gallon)')
plt.ylabel('Price (2013 dollars)')
plt.title('Price vs. mileage')
```

```python
hybrid.iloc[:,1:5].corr()
```

```python
#Conditionally selecting all the rows that correspond to the SUV class

hybrid_suv = hybrid.copy()
hybrid_suv = hybrid_suv[hybrid_suv['class'] == 'SUV']
hybrid_suv
```

```python
#Scatter plot of price vs. acceleration of hybrid SUV cars
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
plt.scatter('acceleration', 'msrp', data = hybrid_suv, color = 'red')
plt.xlabel('Acceleration (Km per hour per second)')
plt.ylabel('Price (2013 dollars)')
plt.title('Price vs. acceleration for SUV class')


#Scatter plot of price vs. mpg of hybrid SUV cars
plt.subplot(1,2,2)
plt.scatter('mpg', 'msrp', data = hybrid_suv, color = 'red')
plt.xlabel('Mileage (miles per gallon)')
plt.ylabel('Price (2013 dollars)')
plt.title('Price vs. mileage for SUV class')
```

```python
#Converting the price in original units to standard units

hybrid_suv['msrp (standard units)'] = (hybrid_suv['msrp'] - np.mean(hybrid_suv['msrp']))/np.std(hybrid_suv['msrp'])
hybrid_suv.head()
```

```python
#Converting the acceleration in original units to standard units

hybrid_suv['acceleration (standard units)'] = (hybrid_suv['acceleration'] - np.mean(hybrid_suv['acceleration']))/np.std(hybrid_suv['acceleration'])
hybrid_suv.head()
```

```python
#Converting the mileage in original units to standard units

hybrid_suv['mpg (standard units)'] = (hybrid_suv['mpg'] - np.mean(hybrid_suv['mpg']))/np.std(hybrid_suv['mpg'])
hybrid_suv
```

```python
#Scatter plot of price vs. acceleration of hybrid SUV cars in standard units
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.scatter('acceleration (standard units)', 'msrp (standard units)', data = hybrid_suv, color = 'red')
plt.xlabel('Acceleration (standard units)')
plt.ylabel('Price (standard units)')
plt.title('Price vs. acceleration for SUV class')
plt.subplot(1,2,2)
plt.scatter('mpg (standard units)', 'msrp (standard units)', data = hybrid_suv, color = 'red')
plt.xlabel('MPG (standard units)')
plt.ylabel('Price (standard units)')
plt.title('Price vs. mileage for SUV class')
```

```python
#Scatter plot of price vs. mileage of hybrid SUV cars in standard units

plt.scatter('mpg (standard units)', 'msrp (standard units)', data = hybrid_suv, color = 'red')
plt.xlabel('MPG (standard units)')
plt.ylabel('Price (standard units)')
plt.title('Price vs. mileage for SUV class')
```

```python
#Printing the first 5 rows of the dataframe 'hybrid_suv'

hybrid_suv.head()
```

```python
#Calculating the correlation coefficient between acceleration and price based on standard units

r = np.mean(hybrid_suv['acceleration (standard units)']*hybrid_suv['msrp (standard units)'])
r
```

```python
#Calculating the correlation coefficient between mileage and price based on standard units

r = np.mean(hybrid_suv['mpg (standard units)']*hybrid_suv['msrp (standard units)'])
r
```

```python
#Using the corrcoef() function to calculate the correlation coefficient between acceleration and price

np.corrcoef(hybrid_suv['acceleration (standard units)'],hybrid_suv['msrp (standard units)'])
```

```python
#Using the corrcoef() function to calculate the correlation coefficient between mileage and price

np.corrcoef(hybrid_suv['mpg (standard units)'],hybrid_suv['msrp (standard units)'])
```

### Limitations of corrleation coefficient r

```python
#Correlation coefficient measures only the 'linear' correlation

x = np.arange(-4,4.1,0.5)
y = x**2

plt.scatter(x, y, color = 'blue')
```

```python
#Using the corrcoef() function to calculate the correlation coefficient

np.corrcoef(x,y)
```

```python
#Correlation coefficient is affected by the outliers

x = [1, 2.1, 3, 4, 5, 6],
y = [1, 2, 2.8, 4, -7, 6]

plt.scatter(x, y, color = 'blue')
```

```python
#Using the corrcoef() function to calculate the correlation coefficient

np.corrcoef(x,y)
```
